```python
import pandas as pd
import numpy as np
import random
from faker import Faker
from datetime import timedelta
```

```python
df = pd.read_csv("orders.csv")

df.head()
```

```python
fake = Faker("en_IN")

products = df["product_name"].unique().tolist()

warehouses = df["warehouse"].unique().tolist()

print(products)

print(warehouses)
```

```python
new_orders = []

for i in range(1, 15001):

    customer_id = random.randint(1000, 9999)

    customer_name = fake.name()

    product_name = random.choice(products)

    warehouse = random.choice(warehouses)

    # Random date between 2022 and 2024
    order_date = fake.date_between(
        start_date="-3y",
        end_date="today"
    )

    expected_delivery_date = (
        order_date +
        timedelta(days=random.randint(3, 7))
    )

    # 80% delivered, 20% delayed
    if random.random() < 0.8:

        actual_delivery_date = (
            expected_delivery_date -
            timedelta(days=random.randint(0, 1))
        )

        status = "Delivered"

        delay_days = 0

    else:

        delay_days = random.randint(1, 15)

        actual_delivery_date = (
            expected_delivery_date +
            timedelta(days=delay_days)
        )

        status = "Delayed"

    new_orders.append({

        "order_id": i,

        "customer_id": customer_id,

        "customer_name": customer_name,

        "product_name": product_name,

        "order_date": order_date,

        "expected_delivery_date":
            expected_delivery_date,

        "actual_delivery_date":
            actual_delivery_date,

        "status": status,

        "delay_days": delay_days,

        "warehouse": warehouse
    })
```

```python
orders_large = pd.DataFrame(new_orders)

orders_large.head()
```
